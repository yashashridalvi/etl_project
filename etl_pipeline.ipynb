ETL :- extract --> transform --> load 
1st step :- extraction

```python
jdbcurl= "jdbc:sqlserver://projectdeserver1.database.windows.net:1433;database=database_de;"
ConnectionProperties={
  "user": "yashashri", 
  "password": "yashRI#5",
  "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver"
}

df = spark.read.jdbc(url=jdbcurl,table= "SalesLT.Customer",properties=ConnectionProperties)

```

```python
display(df)
```

2nd step:- transformation 


```python
#add the timestamp
from pyspark.sql.functions import current_timestamp

df = df.withColumn("current_time", current_timestamp())
display(df)
```

last step of etl is loading


```python
df.write.format('delta').mode('overwrite').saveAsTable('project_data.practice_schema.customer') # here we write or load data in cust
```
