# Config Table

- the table which can store the information about migrating table 

```python
# %sql
# create schema project_data.practice_schema
```

```python
%sql

CREATE SCHEMA IF NOT EXISTS project_data.practice_schema;


CREATE TABLE project_data.practice_schema.etl_migration_config (
  id BIGINT GENERATED ALWAYS AS IDENTITY,
  source_name STRING,
  source_entity_name STRING,
  source_credintials STRING,
  source_type STRING,
  target_table_details STRING,
  created_by TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
  updated_by TIMESTAMP DEFAULT CURRENT_TIMESTAMP
)
USING delta
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');
```
